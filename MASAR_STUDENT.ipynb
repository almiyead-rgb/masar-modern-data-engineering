```python
from pathlib import Path
from contextlib import contextmanager
import json
import subprocess
import sys
ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents)
             if (p / "course.json").is_file() and (p / "src/masar").is_dir()), None)
if ROOT is None:
    raise FileNotFoundError("Open this notebook inside the complete Masar repository")
if str(ROOT / "src") not in sys.path:
    sys.path.insert(0, str(ROOT / "src"))
from masar.runtime import require_environment, start_spark
from masar.workspace import new_workspace, require_fixed_dataset, record_bronze_success
require_environment()
SOURCE = ROOT / "data/masar-small-v1"
require_fixed_dataset(SOURCE)
WORK = new_workspace(ROOT, "student_project")
@contextmanager
def engine(kafka=False):
    session = start_spark(WORK, kafka=kafka)
    try:
        yield session
    finally:
        session.stop()
def show_checks(report):
    checks = report.get("checks", {})
    if not checks or not all(value is True for value in checks.values()):
        raise AssertionError("Inspect the failed native checks: " + str(checks))
    print("Passed checks:", len(checks))
    print("Outputs:", WORK.relative_to(ROOT))
print("Project folder:", WORK.relative_to(ROOT))
```
